```python
# Dify 是 Web 平台，本 notebook 主要讲解概念和 API 调用
# 官方文档: https://docs.dify.ai/

print("""
Dify 核心概念:

1. 应用类型:
   - 聊天助手 (Chatbot):   对话式 AI，支持多轮
   - 文本生成 (Completion): 单次输入→输出
   - Agent:              带工具调用的智能助手
   - 工作流 (Workflow):    可视化拖拽编排

2. 工作流节点:
   - LLM 节点:     调用大模型
   - 知识检索节点:  RAG 检索
   - 代码节点:     执行 Python/JS
   - 条件分支:     IF-ELSE 路由
   - HTTP 请求:    调用外部 API
   - 变量聚合:     合并多个分支结果

3. 知识库:
   - 上传文档 → 自动切分 → Embedding → 向量存储
   - 支持 PDF/Word/Markdown/网页抓取
   - 检索模式: 向量检索 / 全文检索 / 混合检索
""")
```

```python
# Dify API 调用示例
import requests, os

DIFY_API_KEY = os.environ.get("DIFY_API_KEY", "")
DIFY_BASE_URL = "https://api.dify.ai/v1"  # 或自部署地址

def call_dify_chat(query: str, conversation_id: str = "") -> dict:
    """调用 Dify 聊天助手 API"""
    headers = {"Authorization": f"Bearer {DIFY_API_KEY}", "Content-Type": "application/json"}
    payload = {
        "inputs": {},
        "query": query,
        "response_mode": "blocking",
        "conversation_id": conversation_id,
        "user": "test_user",
    }
    try:
        resp = requests.post(f"{DIFY_BASE_URL}/chat-messages", json=payload, headers=headers, timeout=30)
        return resp.json()
    except Exception as e:
        return {"error": str(e)}

print("""
Dify API 模式:
  POST /chat-messages          → 聊天对话
  POST /completion-messages     → 文本生成
  POST /workflows/run           → 运行工作流
  GET  /conversations           → 获取对话列表

B站广告场景:
  1. 知识库: 上传广告投放规范、常见问题文档
  2. 工作流: 用户问题 → 意图识别 → RAG检索 → LLM回答 → 满意度评价
  3. Agent:  绑定广告数据查询工具、文案生成工具
""")
```
