# 07-02 Coze 智能体开发

**Coze**（字节跳动/扣子）是面向 C 端的智能体平台。

**本节目标**：Coze 核心能力、Plugin 机制、知识库、工作流

---

```python
print("""
Coze 核心功能:

1. Bot 创建:
   - 人设 (Persona):     定义角色和行为风格
   - 技能 (Plugins):     内置 600+ 插件（搜索/天气/代码执行等）
   - 知识库 (Knowledge):  上传文档 + 网页抓取
   - 工作流 (Workflow):   可视化编排
   - 记忆 (Memory):       长期记忆 + 变量存储
   - 触发器 (Triggers):   定时任务、Webhook

2. 与 Dify 对比:
   Coze 更偏 C 端（抖音/微信/飞书发布）
   Dify 更偏 B 端（API/SDK/自部署）

3. B站广告场景 Bot 设计:
   人设: B站广告投放顾问
   知识库: 广告政策文档 + FAQ
   插件: 广告数据查询API + 文案生成
   工作流: 问题分类 → 检索知识库 → 生成回答 → 满意度追问
""")
```
